Partie 5 – Manipulation des colonnes
1) Ajouter la colonne "temperature_fahrenheit"
2) Ajouter la colonne "niveau_temperature"
 pour catégoriser la température en  
"Élevée" (température > 30) ou "Normale". On peut utiliser la méthode where() de numpy
3) Renommer la colonne "humidite" en "humidite_relative"
4) Supprimer la colonne "temperature_fahrenheit"

```python
# 1) Ajouter la colonne "temperature_fahrenheit"
df["temperature_fahrenheit"] = df["temperature"] * 9/5 + 32
df.head()
```

```python
# 2) Ajouter la colonne "niveau_temperature" (catégorisation avec np.where)
df["niveau_temperature"] = np.where(df["temperature"] > 30, "Élevée", "Normale")
df.head()
```

```python
# 3) Renommer la colonne "humidite" en "humidite_relative"
df = df.rename(columns={"humidite": "humidite_relative"})
df.head()
```

```python
# 4) Supprimer la colonne "temperature_fahrenheit"
df = df.drop(columns=["temperature_fahrenheit"])
df.head()
```

Partie 6 – Filtrage
1) Récupérer uniquement les lignes telles que température
 > 30
2) Récupérer uniquement les lignes telles température
 > 30 et humidité > 70

```python
# 1) Lignes où température > 30
df_temp_sup_30 = df[df["temperature"] > 30]
df_temp_sup_30
```

```python
# 2) Lignes où température > 30 ET humidité > 70
df_filtre = df[(df["temperature"] > 30) & (df["humidite_relative"] > 70)]
df_filtre
```

Partie 9 : Gestion des valeurs manquantes
1) Afficher le nombre de valeurs manquantes par colonne
2) Afficher le nombre total des valeurs manquantes
3) Afficher le taux de valeurs manquantes
4) Afficher les lignes contenant des valeurs manquantes
5) Pour la température et l’humidité, remplacer les valeurs manquantes par les moyennes
6) Pour la consommation, remplacer les valeurs manquantes par la médiane

```python
# 1) Nombre de valeurs manquantes par colonne
df.isnull().sum()
```

```python
# 2) Nombre total de valeurs manquantes
df.isnull().sum().sum()
```

```python
# 3) Taux de valeurs manquantes
taux_manquant = df.isnull().sum().sum() / df.size * 100
taux_manquant
```

```python
# 4) Lignes contenant des valeurs manquantes
df[df.isnull().any(axis=1)]
```

```python
# 5) Remplacer les valeurs manquantes de température et humidité par la moyenne
df["temperature"] = df["temperature"].fillna(df["temperature"].mean())
df["humidite_relative"] = df["humidite_relative"].fillna(df["humidite_relative"].mean())
```

```python
# 6) Remplacer les valeurs manquantes de consommation par la médiane
df["consommation"] = df["consommation"].fillna(df["consommation"].median())
```

```python
# 7) Remplacer les valeurs manquantes de "etat" par "INCONNU"
df["etat"] = df["etat"].fillna("INCONNU")
```
