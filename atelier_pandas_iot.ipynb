Partie 1 – Series
1) Créer une Series contenant 4 températures  
2) Afficher son index
3) Afficher la 1ère valeur
4) Afficher la dernière valeur
5) Créer, de deux façon différentes, une Series de 4 températures avec comme index les heures de
mesure 12h, 13h, 14h et 15h
6) Afficher la 1ère et la dernière valeur

```python
import pandas as pd
import numpy as np
```

```python
# 1) Créer une Series contenant 4 températures
temperatures_series = pd.Series([24.5, 25.1, 26.3, 27.0])
temperatures_series
```

```python
# 2) Afficher son index
temperatures_series.index
```

```python
# 3) Afficher la 1ère valeur
temperatures_series[0]
```

```python
# 4) Afficher la dernière valeur
temperatures_series.iloc[-1]
```

```python
# 5) Series avec index personnalisé (heures), deux méthodes

# Méthode 1 : en passant le paramètre index directement
temperatures_horaires_v1 = pd.Series([24.5, 25.1, 26.3, 27.0], index=["12h", "13h", "14h", "15h"])
temperatures_horaires_v1
```

```python
# Méthode 2 : en créant d'abord la Series par défaut, puis en modifiant .index
temperatures_horaires_v2 = pd.Series([24.5, 25.1, 26.3, 27.0])
temperatures_horaires_v2.index = ["12h", "13h", "14h", "15h"]
temperatures_horaires_v2
```

```python
# 6) Afficher la 1ère et la dernière valeur
print("1ère valeur   :", temperatures_horaires_v1.iloc[0])
print("Dernière valeur:", temperatures_horaires_v1.iloc[-1])
```
